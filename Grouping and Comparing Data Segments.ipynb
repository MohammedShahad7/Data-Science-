```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df=sns.load_dataset('tips')

df.head()
```

```python
df['Spending_Type']=np.where(df['total_bill']>20,"Premium","Budget")
```

```python
df['tip_pct']=(df['tip']/df['total_bill']*100)
```

```python
df[['Spending_Type','tip_pct']].head()
```

```python
summart = pd.pivot_table(
    df,
    values=['tip_pct','size'],
    index=['day','Spending_Type'],
    aggfunc={
        'tip_pct':'median',
        'size':'mean'
    }
)

print(summart.round(2))
```

```python
sammer=df.groupby(['day','Spending_Type']).agg({
    'tip_pct':'median',
    'size':'mean'
}).round(2)

print(sammer)
```

```python
global_avg_tip = df['tip_pct'].mean()

global_avg_tip
```

```python
avg_tip_by_type = df.groupby('Spending_Type')['tip_pct'].mean()

avg_tip_by_type
```

```python
delta=avg_tip_by_type['Budget']- avg_tip_by_type['Premium']

delta
```
