```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df_load=sns.load_dataset('penguins')

print(df_load.head())
print('\n',df_load.shape)
print('\n',df_load.columns)
print('\n',df_load.isna().sum())
```

```python
url='https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv'

df_git=pd.read_csv(url)

print(df_git.head())
print('\n',df_git.shape)
print('\n',df_git.columns)
print('\n',df_git.isna().sum())
```
