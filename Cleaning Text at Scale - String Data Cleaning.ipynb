```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)
```

```python
df.head()
```

```python
# Observation: I expect numeric-only tickets to produce NaN, while tickets with letters or symbols will produce a prefix.

df['ticket_prefix'] = df['Ticket'].str.extract(r'^([^0-9]+)')
```

```python
df['ticket_prefix'].isna().sum()
```

```python
df['ticket_prefix']=df['ticket_prefix'].str.strip().str.upper()
```

```python
df['ticket_prefix']=df['ticket_prefix'].fillna('Numeric')
```

```python
df['ticket_prefix']=df['ticket_prefix'].str.replace('.','').str.replace('/','')
```

```python
df['ticket_prefix'
]
```

```python
group=df.groupby('ticket_prefix')['Survived'].agg(
    survival_rate='mean',
    passenger_count='count'
).sort_values(
        'survival_rate',
        ascending=False
    )

group
```
