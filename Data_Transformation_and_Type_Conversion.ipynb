```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv")  # replace with your dataset

# Categorical columns (object type)
categorical_cols = df.select_dtypes(include='object').columns

# Numerical columns (int + float)
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

print("Categorical Columns:")
print(list(categorical_cols))

print("\nNumerical Columns:")
print(list(numerical_cols))
```

```python
df['total_bill'] = df['total_bill'].astype('float64')
df['tip'] = df['tip'].astype('float64')

# Enforce Integer
df['size'] = df['size'].astype('int64')

# Convert to String (object)
df['day'] = df['day'].astype('object')

# Verification
print(df.dtypes)
```

```python


# Labeling: sex column (Male → M, Female → F)
df['sex'] = df['sex'].map({'Male': 'M', 'Female': 'F'})

# Logic Casting: smoker column (Yes → True, No → False)
df['smoker'] = df['smoker'].map({'Yes': True, 'No': False})

# Verification using value_counts()
print("Sex column value counts:")
print(df['sex'].value_counts())

print("\nSmoker column value counts:")
print(df['smoker'].value_counts())
```

```python
df['time'] = df['time'].str.lower().str.strip()

# Display unique values
print("Unique values in 'time' column:")
print(df['time'].unique())
```
