```python
import pandas as pd
```

```python
URL="https://raw.githubusercontent.com/CodeVinayak/Student-Performance-Analysis-Machine-Learning/master/StudentsPerformance.csv"

df=pd.read_csv(URL)

df.head()
```

```python
scores_subset=df[['reading_score','math_score','gender']]

scores_subset.head()
```

```python
The_10_19=df.iloc[10:20]
print("The row 10 t0 19:\n",The_10_19)
```

```python
only_MAth=df['math_score']
print("Math Score:\n",only_MAth)
```

```python
The_math_score_more_then_80=df[df['math_score']>80]
print("The math score more then 80:\n",The_math_score_more_then_80)
```

```python
The_read_75__write_70=df[(df['reading_score']>75)&(df['writing_score']>70)]
print("The read score more then 75 & write score more then 70:\n 5 ROW: \n",The_read_75__write_70.head())
```
