```python
import pandas as pd
import numpy as np

data = {
    'Customer_ID': [1, 2, 3, 4, 5, 6, 8, 9, 10],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi', 'Ivan'],
    'Age': [25, np.nan, 30, 22, 45, np.nan, 33, 38, np.nan],
    'City': ['New York', 'London', np.nan, 'London', 'Paris', 'New York', 'London', np.nan, 'Paris'],
    'Spend_Amount': [150, 200, 120, np.nan, 300, 180, np.nan, 210, 190],
    'Feedback_Notes': [np.nan, np.nan, np.nan, np.nan, np.nan, np.nan, np.nan, np.nan, np.nan]
}

df = pd.DataFrame(data)

df.head()
```

```python
df.info()
```

```python
num_row=df.shape[0]
print("Number of Row: \n",num_row)
num_col=df.shape[1]
print("Number of Column: \n",num_col)
```

```python
all_is_null = df.columns[df.isnull().all()][0]
print("Columns where all values are null:", all_is_null)
```

```python
print("The `Notes` column should be dropped because it contains 100% missing values (all entries are null). Since it has no useful data, keeping it would only waste memory and add unnecessary complexity to the dataset.")
df.dropna(axis=1,how='all',inplace=True)
df.head()
```

```python
avg_Age=df['Age'].mean()
df['Age'].fillna(avg_Age,inplace=True)
df.head()
```

```python
mode_City=df['City'].mode()[0]
df['City'].fillna(mode_City,inplace=True)
df.head()
```

```python
df.dropna(axis=0,subset=['Spend_Amount'],inplace=True)
df.head()
```
