# Subtask 4: Tidy Violation Summary

### 1. Which of the three tidy data rules did the Titanic dataset violate?

The Titanic dataset does **not have a major violation of the three tidy data rules**.

* **Rule 1: Each variable forms exactly one column.**
  The dataset mostly satisfies this rule because variables such as `age`, `sex`, `fare`, `embarked`, `deck`, and `survived` have their own columns.

* **Rule 2: Each observation forms exactly one row.**
  The dataset satisfies this rule because each row represents information about **one passenger**.

* **Rule 3: Each type of observational unit forms exactly one table.**
  The dataset satisfies this rule because the table describes one main observational unit: **passengers**.

Therefore, **no specific column clearly violates the three core tidy-data rules in this version of the Titanic dataset**.

### 2. Which rules did it satisfy cleanly?

The dataset satisfies all three tidy data rules:

1. **One variable per column** – variables are stored in separate columns.
2. **One observation per row** – each row represents one passenger.
3. **One type of observational unit per table** – the table contains passenger-level information.

The dataset may still contain missing values and other data-quality issues, but missing values do not automatically mean that the data is not tidy.

### 3. One data quality issue that today's operations cannot fix

One important data-quality issue is the **missing values**, especially in the `age` and `deck` columns.

The operations used today can **identify** missing values using:

```python
df.isnull().sum()
```

but they cannot decide what the missing value should actually be.

To fix this, we need a **data-imputation or domain-based decision**. For example, missing `age` values could be replaced using the median age or another suitable statistical method. Missing `deck` values could potentially be left as missing if there is no reliable information available.

This requires a decision based on the meaning of the data, not just a simple cleaning operation.
