```python
import pandas as pd
```

```python
energy_usage_houseA=pd.Series([12.5, 13.2, 11.8, 14.0, 15.1],index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri'])

print("Energy usage of house A:")
print(energy_usage_houseA)
```

```python
energy_usage_on_wed=energy_usage_houseA.loc["Wed"]
print("Energy usage on Wednesday:")
print(energy_usage_on_wed)
```

```python
energy_usage_houseB=pd.Series({'Mon': 10.2,
  'Tue': 11.5,
  'Thu': 12.0,
  'Fri': 13.3})
print("Energy usage of house B:")
print(energy_usage_houseB)
```

```python
total_consumption =energy_usage_houseA+energy_usage_houseB
print("Total consumption:")
print(total_consumption)
```
